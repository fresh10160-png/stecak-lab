# Analiza motiva i natpisa

Učestalost ornamentalnih motiva kroz katalog te udio nekropola s dokumentiranim epigrafskim natpisima.

```python
import sys
sys.path.append('..')
import matplotlib.pyplot as plt

from stecak_lab.data import load_dataframe, stats_motiva, stats_po_drzavi

df = load_dataframe()
motivi = stats_motiva(df)
motivi
```

## Grafikon učestalosti motiva

```python
fig, ax = plt.subplots(figsize=(8, max(3, len(motivi) * 0.35)))
motivi.sort_values().plot.barh(ax=ax, color='#4f653d')
ax.set_xlabel('Broj lokaliteta')
ax.set_ylabel('')
ax.set_title('Učestalost motiva u katalogu')
fig.tight_layout()
plt.show()
```

## Broj stećaka po državi

```python
po_drzavi = stats_po_drzavi(df)

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(po_drzavi['drzava'], po_drzavi['broj_stecaka'], color='#65804e')
ax.set_ylabel('Broj stećaka (okvirno)')
ax.set_title('Broj stećaka po državi')
fig.autofmt_xdate(rotation=20)
fig.tight_layout()
plt.show()
```

## Udio nekropola s dokumentiranim natpisima
